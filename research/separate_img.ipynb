```python
import os
import shutil

# Source folder
source_folder = r"C:\Users\Nagham\whatever\Test_Images"

# Destination folders
dest_folder_1 = os.path.join(source_folder, "folder1")
dest_folder_2 = os.path.join(source_folder, "folder2")
dest_folder_3 = os.path.join(source_folder, "folder3")
dest_folder_4 = os.path.join(source_folder, "folder4")
dest_folder_5 = os.path.join(source_folder, "folder5")
dest_folder_6 = os.path.join(source_folder, "folder6")
dest_folder_7 = os.path.join(source_folder, "folder7")


# Create destination folders if they don't exist
os.makedirs(dest_folder_1, exist_ok=True)
os.makedirs(dest_folder_2, exist_ok=True)
os.makedirs(dest_folder_3, exist_ok=True)
os.makedirs(dest_folder_4, exist_ok=True)
os.makedirs(dest_folder_5, exist_ok=True)
os.makedirs(dest_folder_6, exist_ok=True)
os.makedirs(dest_folder_7, exist_ok=True)


# Get list of all image files in the source folder
image_files = [f for f in os.listdir(source_folder) if os.path.isfile(os.path.join(source_folder, f))]

# Separate images into three folders
for i, file in enumerate(image_files):
    if i % 7 == 0:
        shutil.move(os.path.join(source_folder, file), os.path.join(dest_folder_1, file))
    elif i % 7 == 1:
        shutil.move(os.path.join(source_folder, file), os.path.join(dest_folder_2, file))
    elif i % 7 == 2:
        shutil.move(os.path.join(source_folder, file), os.path.join(dest_folder_3, file))
    elif i % 7 == 3:
        shutil.move(os.path.join(source_folder, file), os.path.join(dest_folder_4, file))
    elif i % 7 == 4:
        shutil.move(os.path.join(source_folder, file), os.path.join(dest_folder_5, file))
    elif i % 7 == 5:
        shutil.move(os.path.join(source_folder, file), os.path.join(dest_folder_6, file))
    else:
        shutil.move(os.path.join(source_folder, file), os.path.join(dest_folder_7, file))

print("Images have been separated into 7 folders.")
```
