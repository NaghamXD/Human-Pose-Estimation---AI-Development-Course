```python
import os
from PIL import Image
```

```python
def load_bounding_boxes(file_path):
    """Load bounding box data from a text file."""
    bounding_boxes = []
    with open(file_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            class_id = int(parts[0])
            center_x = float(parts[1])
            center_y = float(parts[2])
            width = float(parts[3])
            height = float(parts[4])
            bounding_boxes.append((class_id, center_x, center_y, width, height))
    return bounding_boxes

def crop_image(image, bbox):
    """Crop the image based on the bounding box."""
    img_width, img_height = image.size
    class_id, center_x, center_y, width, height = bbox
    
    left = (center_x - width / 2) * img_width
    top = (center_y - height / 2) * img_height
    right = (center_x + width / 2) * img_width
    bottom = (center_y + height / 2) * img_height
    
    return image.crop((left, top, right, bottom))

def resize_image(image, size=(384, 288)):
    """Resize the image to the given size."""
    return image.resize(size, Image.LANCZOS)

def list_images(image_dir):
    """List all images in the directory."""
    return [f for f in os.listdir(image_dir) if os.path.isfile(os.path.join(image_dir, f))]

def process_images(image_dir, label_dir, output_dir, size=(384, 288)):
    """Process images by cropping and resizing based on bounding box labels."""
    images = list_images(image_dir)
    for label_file in os.listdir(label_dir):
        if label_file.endswith('.txt'):
            image_prefix = label_file[:7]
            image_path = None
            
            for img in images:
                if img.startswith(image_prefix) and (img.endswith('.jpg') or img.endswith('.png')):
                    image_path = os.path.join(image_dir, img)
                    break
            
            if image_path:
                image = Image.open(image_path)
                label_path = os.path.join(label_dir, label_file)
                bounding_boxes = load_bounding_boxes(label_path)
                
                for i, bbox in enumerate(bounding_boxes):
                    cropped_image = crop_image(image, bbox)
                    resized_image = resize_image(cropped_image, size)
                    
                    output_path = os.path.join(output_dir, f"{label_file[:-4]}_bbox_{i}.jpg")
                    resized_image.save(output_path)
                    print(f"Processed {output_path}")
            else:
                print(f"Image with prefix {image_prefix} not found.")
```

```python
image_directory = r'C:\Users\Nagham\whatever\Labeled Images with Roboflow'
label_directory = r'C:\Users\Nagham\whatever\Labels From Roboflow'
output_directory = r'C:\Users\Nagham\Human-Pose-Estimation---AI-Development-Course\dataset\croped_images'

process_images(image_directory, label_directory, output_directory)
```
