```python
# Define paths
model_path = r"/home/usera1/Human-Pose-Estimation---AI-Development-Course/yolov5/runs/train/exp8/weights/best.pt"
image_folder = r"/home/usera1/Human-Pose-Estimation---AI-Development-Course/dataset/images/folder3"
output_folder = r"/home/usera1/Human-Pose-Estimation---AI-Development-Course/dataset/labels/test/folder3_7"

# Load the model
model = torch.hub.load('./yolov5', "custom", path=model_path, source="local")  # local repo

# Create the output folder if it doesn't exist
Path(output_folder).mkdir(parents=True, exist_ok=True)

# Iterate over all images in the folder
for image_path in Path(image_folder).glob('*.*'):  # Matching all image formats
    # Perform inference
    results = model(str(image_path))

    # Save the image with bounding boxes
    results.save(Path(output_folder) / image_path.name)

    # Extract bounding box data and save it
    bbox_data = []
    for *box, conf, cls in results.xyxy[0].tolist():
        bbox_data.append({
            'bbox': box,
            'confidence': conf,  # Use directly, already a float
            'class': int(cls)    # Use directly, already an int
        })

    # Save bounding box data to a JSON file
    with open(Path(output_folder) / f"{image_path.stem}_bboxes.json", 'w') as f:
        json.dump(bbox_data, f, indent=4)

print('Inference completed.')
```
